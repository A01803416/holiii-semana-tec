```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file_path = 'recipes_extended.csv' 
df = pd.read_csv(file_path)
```

```python
df.head()
```

```python
df.columns
```

```python
df['ingredients']
```

```python
health_counts = df['health_level'].value_counts()

print(health_counts)
```

```python
plt.figure(figsize=(8, 5))
```

```python
plt.title('Clasificación de Recetas por Nivel de Salud', fontsize=14)
plt.ylabel('Cantidad de Recetas', fontsize=10)
plt.bar(
    health_counts.index, 
    health_counts.values,
    color=["#2266ba", "#c77529"]
)
```
